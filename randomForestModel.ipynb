Loading the Dataset

```python
import pandas as pd

train_url = "https://raw.githubusercontent.com/tikadatta2005/ML_GROUP_ASSIGNMENT/refs/heads/main/datasets/train.csv"
test_url = "https://raw.githubusercontent.com/tikadatta2005/ML_GROUP_ASSIGNMENT/refs/heads/main/datasets/test.csv"

print("Loading data directly from GitHub...")
train_df = pd.read_csv(train_url)
test_df = pd.read_csv(test_url)
print("Data loaded successfully!")

train_df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_train = le.fit_transform(train_df['dengue_total'])
y_test = le.transform(test_df['dengue_total'])

X_train = train_df.drop(columns=['dengue_total', 'Location', 'Month'])
X_test = test_df.drop(columns=['dengue_total', 'Location', 'Month'])

print("--- Step 2 Complete ---")
print("Features left for training:", X_train.columns.tolist())
print("Target classes successfully converted to numbers!")
```

Random Forest Implementation

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
import numpy as np

print("Bucketing raw case numbers into categories and training... Please wait.")

def convert_to_risk_level(case_count):
    if case_count <= 5:
        return 'Low Risk'
    elif case_count <= 50:
        return 'Medium Risk'
    else:
        return 'High Risk'

y_train_categorical = train_df['dengue_total'].apply(convert_to_risk_level)
y_test_categorical = test_df['dengue_total'].apply(convert_to_risk_level)

X_train_forced = train_df.drop(columns=['dengue_total', 'Location', 'Month'], errors='ignore')
X_train_forced = X_train_forced.select_dtypes(include=[np.number])

X_test_forced = test_df.drop(columns=['dengue_total', 'Location', 'Month'], errors='ignore')
X_test_forced = X_test_forced.select_dtypes(include=[np.number])

my_rf_model = RandomForestClassifier(n_estimators=100, max_depth=8, class_weight='balanced', random_state=42)
my_rf_model.fit(X_train_forced, y_train_categorical)
print("Training complete!")

y_pred = my_rf_model.predict(X_test_forced)

print("\n--- YOUR OFFICIAL ASSIGNMENT RESULTS ---")
print(f"Accuracy Score: {accuracy_score(y_test_categorical, y_pred):.4f}")
print("\nDetailed Performance Report:")
print(classification_report(y_test_categorical, y_pred, zero_division=0))
```
